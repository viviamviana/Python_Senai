# Aula 02 - Tipos de dados

```python
# Tipos numéricos - Inteiros e Decimais (int | float)

x = 10     # Inteiro
y = -1     # Inteiro negativo
pi = 3.14  # Float
z = -5.29  # Float negativo

print(type(x))
print(type(y))
print(type(pi))
print(type(z))
```

```python
# Tipo Caracter - String (str)

nome = 'Viviam'
escola = "SENAI Ary Torres"

print(nome)
print(escola)
print()
print(type(nome)) # Mostra o tipo da variável
print('\n', escola) # '\n' pula uma linha
print('\n', escola[2]) # Retorna a letra da posição 2
```

```python
# Tipo Lógico - Booleano (bool)

continuar = True
parar = False
casado = True

print(continuar)
print(parar)
print(type(casado))
```

```python
# Declaração de variáveis

texto   = "2"      # Declaração de variável string
texto1  = str(2)   # Declaração de variável do tipo String
numero  = 3        # Declaração de variável numérica, podendo ser inteiro ou decimal
numero1 = int(3)   # Declaração de variável numérica, do tipo inteiro
numero2 = float(3) # Declaração de variável numérica, do tipo decimal
logico  = bool(0)  # Atribuição de variável lógica, do tipo booleana (0:false / 1:true)

print(texto)
print(texto1)
print(numero)
print(numero1)
print(numero2)
print(logico)
```

```python
# Ler valores inseridos pelo usuário - Entrada de dados

# Atribui dados digitados pelo usuário e converte em Número Inteiro
numeroInteiro = int(input("Digite um número inteiro: "))

# Atribui dados digitados pelo usuário e converte em Número Decimal
numeroReal = float(input("Digite um número Decimal: "))

# Atribui dados digitados pelo usuário e converte em Bool
booleano = bool(input("Digite um valor Booleano: "))

# Atribui dados digitados pelo usuário e converte em Texto
string = str(input("Digite um texto: "))
```

```python
# Identificar o tipo primitivo

print(type(numeroInteiro))
print(type(numeroReal))
print(type(booleano))
print(type(string))
```

```python
# Cores - Tabela ANSI

"""
 "\033[ESTILO;FONTE;FUNDOm TextoMostrado \033[0m"

 - Estilo: 0 nada, 1 negrito, 2 itálico, 4 underline e 7 negativo
 - Fonte: 30 ao 37
 - Fundo: 40 ao 47
 - \033[0m no final para de estilizar o texto
"""

print("\033[1;33;45mtexto\033[0m")

print("\033[30m Cinza")
print("\033[31m Vermelho")
print("\033[32m Verde")
print("\033[33m Amarelo")
print("\033[34m Ciano")
print("\033[35m Magenta")
print("\033[36m Azul")
print("\033[37m Branco")
```

```python
# Colunas com preenchimento e alinhamento

print(" CENTRALIZADO ".center(20,'-')) # Centraliza, em coluna, de acordo com o número  de caracteres.
print(" DIREITA".rjust(20,'-')) # Alinha à direita
print("ESQUERDA ".ljust(20,'-'))# Alinha à esquerda
```

```python
print("CENTRALIZADO".center(20))
print("DIREITA".rjust(20))
print("ESQUERDA".ljust(20))
```

```python
# Alinhamento para f-strings

print(f"| {'Unidade':>20} | {'5':^20} | ") # a palavra 'unidade' vai ser alinhada em 20 colunas. O ">" serve para indicar que será alinhada à direita.
print(f"| {'Unidade':^20} | {'5':^20} | ") # o "^" alinha ao centro.
print(f"| {'Unidade':<20} | {'5':^20} | ") # o "<" alinha à esquerda.
```

```python
# Multiplicação de strings

hashtag = '#'
asterisco = '*'

print(hashtag*2, asterisco*10, hashtag*2)
print('='*15)
```

## Desafio 09

```python
# DESAFIO 09
"""
Crie um programa e declare uma constante PI utilizando 4 casas
decimais. Dados o raio e a altura, calcule e apresente o volume de um
objeto cilíndrico.
Fórmula: volume = PI * r² * altura
"""
PI = 3.1415
r = float(input("Insira o valor do raio: "))
h = float(input("Insira o valor da altura: "))

vol = PI * r**2 * h

print(f"\nO volume do cilíndro é: {vol:,.2f}")
```
