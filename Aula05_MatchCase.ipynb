# Aula 05 - Match Case

```python
# Estrutura Match Case
operacao = input("""
ESCOLHA UMA OPÇÃO:

[1] Opção 1
[2] Opção 2
[3] Opção 3
""")

match operacao: # match = corresponde
    case '1': # case = caso
        print("Opção 1 foi selecionada.")
    case '2': # as aspas servem para indicar que é uma string
        print("Opção 2 foi selecionada.")
    case '3':
        print("Opção 3 foi selecionada.")
    case _: # serve como um else
        print("Opção inválida...")
```

```python
# Mútipla operação
operacao = input("""
ESCOLHA UMA OPÇÃO:

[1] Opção 1
[2] Opção 2
[3] Opção 3
""")

match operacao:
    case '1' | '2': # a '|' equivale a um 'or' dentro do match case.
        print("Opção 1 ou 2 foi selecionada.")
    case '3':
        print("Opção 3 foi selecionada.")
    case _:
        print("Opção inválida...")
```

```python
# Múltipla operação com case _ if
numero = int(input("Digite um número: "))

match numero:
    case 1 | 3 | 5 | 7 | 9:
        print("É ímpar entre 0 e 10")
    case _ if numero < 0:
        print("É negativo")
    case _:
        print("Um número qualquer")
```

## Desafio 01

```python
# DESAFIO 01
"""
Escreva um programa em Python que solicite ao usuário uma nota entre 0 e 10.
Use o comando 'match case' para classificar a nota de acordo com a tabela a seguir:

- 0-4: Nota Baixa
- 5-7: Nota Média
- 8-9: Nota Alta
- 10: Nota Excelente

Exiba a classificação correspondente.
"""

nota = int(input("Digite uma nota de 0 a 10: "))

match nota:
    case _ if nota <= 4:
        print("Nota baixa")
    case _ if nota <= 7:
        print("Nota média")
    case _ if nota <= 9:
        print("Nota alta")
    case _ if nota <= 10:
        print("Nota excelente")
```

```python
# DESAFIO 01 - V2

nota = int(input("Digite uma nota de 0 a 10: "))

match nota:
    case 0 | 1 | 2 | 3 | 4:
        print("Nota baixa")
    case 5 | 6 | 7:
        print("Nota média")
    case 8 | 9:
        print("Nota alta")
    case _:
        print("Nota excelente")
```

```python
# DESAFIO 01 - V3

nota = int(input("Digite sua nota: "))

match nota:
    case 0 | 1 | 2 | 3 | 4:
        print(f"\nNota Baixa: {nota}")
    case 5 | 6 | 7:
        print(f"\nNota Média: {nota}")
    case 8 | 9:
        print(f"\nNota Alta: {nota}")
    case 10:
        print(f"\nNota Excelente: {nota}")
    case _:
        print("Nota inválida...")
```

## Desafio 03

```python
# DESAFIO 03
"""
Implemente uma calculadora simples que realiza as operações de adição, subtração, multiplicação e divisão.
O usuário deve inserir dois números e, em seguida, selecionar a operação desejada (+, -, *, /).
Use o comando match case para realizar a operação correta e exibir o resultado.
"""

num1 = float(input("Insira o 1° número: "))
num2 = float(input("Insira o 2° número: "))
operacao = input("""
Escolha uma operação:
[1] Adição
[2] Subtração
[3] Multiplicação
[4] Divisão

""")

match operacao:
    case '1':
        print(f"\nValor da soma entre os dois números: {num1+num2}")
    case '2':
        print(f"\nValor da subtração entre os dois números: {num1-num2}")
    case '3':
        print(f"\nValor da multiplicação entre os dois números: {num1*num2}")
    case _ if operacao == '4' and num2 != 0: # não se divide por 0
        print(f"\nValor da divisão entre os dois números: {num1/num2:,.1f}")
    case _:
        print("\nNúmero inválido...")
```

```python
# DESAFIO 03 - V2

num1 = float(input("Digite o 1º número: "))
num2 = float(input("Digite o 2º número: "))
operacao = input("Selecione a operação: (+ - * /): ")

match operacao:
    case '+':
        print(num1+num2)
    case '-':
        print(num1-num2)
    case '*':
        print(num1*num2)
    case '/':
        if num2 != 0:
            print(num1/num2)
        else:
            print("Não divisível por zero...")
    case _:
        print("Operação inválida")
```

## Desafio 04

```python
# DESAFIO 04
"""
Escreva um programa que converta valores entre diferentes moedas.
O usuário deve inserir o valor em reais e selecionar a moeda para conversão:
D (Dólar), E (Euro), ou L (Libra). Utilize match case para aplicar a conversão
correta com base nas taxas fictícias fornecidas a seguir:

1 Real = 0.18 Dólar
1 Real = 0.16 Euro
1 Real = 0.13 Libra

Símbolos: $ € £
"""

r = float(input("Insira o valor em R$: "))
moeda = input("""
Escolha a moeda para conversão:
[d] Dólar
[e] Euro
[l] Libra

""")

match moeda:
    case 'd':
        print(f"\nR${r} convertido para ${(r*0.18):,.2f}")
    case 'e':
        print(f"\nR${r} convertido para €{(r*0.16):,.2f}")
    case 'l':
        print(f"\nR${r} convertido para £{(r*0.13):,.2f}")
```

```python
# DESAFIO 04 - V2

r = float(input("Insira o valor em R$: "))
moeda = input("""
Escolha a moeda para conversão:
[d] Dólar
[e] Euro
[l] Libra

""")

match moeda:
    case 'd' | 'D':
        print(f"\nR${r} convertido para ${(r*0.18):,.2f}")
    case 'e' | 'E':
        print(f"\nR${r} convertido para €{(r*0.16):,.2f}")
    case 'l' | 'L':
        print(f"\nR${r} convertido para £{(r*0.13):,.2f}")
    case _:
        print("Opção inválida")
```

```python
# DESAFIO 04 - V3

valor = float(input("Digite um valor em R$ para conversão: "))
conversao = input("Escolha a moeda: D (Dólar) E (Euro) L (Libra) ").lower()

match conversao:
    case 'd':
        print(f"$ {(valor*0.18):,.2f}")
    case 'e':
        print(f"€ {(valor*0.16):,.2f}")
    case 'l':
        print(f"£ {(valor*0.13):,.2f}")
    case _:
        print("Conversão inválida...")
```

## Desafio 05

```python
# DESAFIO 05
"""
Crie um programa que peça ao usuário para inserir uma letra. O programa deve utilizar match case
para verificar se a letra é uma vogal (a, e, i, o, u) ou uma consoante, e então exibir uma mensagem apropriada.
"""

letra = input("Digite uma letra: ").lower()

match letra:
    case 'a' | 'e' | 'i' | 'o' | 'u':
        print("A letra digitada é uma vogal!")
    case _:
        if letra.isalpha(): # O teste lógico tá verificando se o que foi digitado é uma letra.
            print("A letra digitada é uma consoante!")
        else:
            print("Opção inválida...")
```

```python
# DESAFIO 05 - V2

letra = input("Digite uma letra: ").lower()

match letra:
    case "a" | "e" | "i" | "o" | "u":
        print(f"{letra} é uma vogal.")
    case _ if letra.isalpha():
        print(f"{letra} é uma consoante.")
    case _:
        print("Caractere inválido")
```
