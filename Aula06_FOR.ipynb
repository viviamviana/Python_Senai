# Aula 06 - Estrutura de repetição FOR

```python
# Laço FOR - Exemplo 01
for i in range(1, 6, 2): # range(início, fim, salto)
    print(i)

print()

# Exemplo 02
for j in range(0, 10, 2):
    print(j, end=' ') # O "end=''" permite você substituir a quebra por qualquer outra coisa dentro das aspas.
# cada print, por padrão, contém uma quebra de linha.
```

```python
# Exemplo 03
for k in range(1, 6):
# sem definir o salto, ele segue de um em um.
    print(k)

print()

# Exemplo 04
for l in range(1, 6):
    print(l, end=' ')
```

```python
# Exemplo 05
inicio = 5
fim = -1
salto = -1 # decrescendo de um em um

for i in range(inicio, fim, salto):
    print(i)

print()

# Exemplo 06
for j in range(inicio, fim, salto):
    print(j, end=' ')
```

```python
# Exemplo 07
for i in range(10):
    print(f"{i} | SENAI")
```

```python
# Exemplo 08
for i in range(10):
    print(f"{i+1} | SENAI") # +1 para começar do 1
```

```python
# Exemplo 09
texto = "PYTHON"
for letra in texto:
    print(letra)

print()

texto = "PYTHON"
for letra in texto:
    print(letra, end=' ')
```

```python
# Exemplo 10
frase = "PYTHON É FANTÁSTICO"
for palavra in frase:
    print(palavra, end='-')

print()

# Exemplo 11
frase = ['Python', 'é', 'fantástico!']

for palavra in frase:
    print(palavra, end=' ')
```

```python
# Exemplo 12
for i in range(50):
    if i >= 5:
        print("Acabou o laço")
        break # faz a quebra do laço
    print(i)
```

```python
# Exemplo 13
lista = [1, 10, 20, 30, 40, 50] # a lista permite declarar mais de uma variável dentro dela.
for num in lista:
    if num == 30:
        continue # ignora o if e continua executando.
    print(num, end=' ')
```

```python
# Exemplo 14
for i in range(1, 11):
    print()
    for j in range(1, 11):
        print(f"{i*j}".rjust(3), end=' ')
```

```python
print("04", "09", "2024", sep='/') # separa com o que tiver entre ''
```

## Desafio 03

```python
# DESAFIO 03
"""
Faça um programa que calcule a soma entre todos os números ímpares
que são múltiplos de três e que se encontram no intervalo de 1 até 500.
"""

soma = 0

for num in range(0, 501):
    if num%3 == 0 and num%2 != 0:
        soma += num
print(soma)
```

```python
# DESAFIO 03 - V3

soma = 0
for i in range(3, 500, 2):
    if i % 3 == 0:
        soma += i
print(f"Soma: \033[1m{soma}\033[0m")
```

```python
# DESAFIO 03 - V4

soma = 0
for num in range(0, 501, 3):
    if num%2 != 0:
        soma+=num
soma = f"{soma:,.0f}"
soma = soma.replace(',', '.')
print(soma)
```

## Desafio 07

```python
# DESAFIO 07
"""
Crie um programa que leia o ano de nascimento de sete pessoas. No final,
mostre quantas pessoas ainda não atingiram a maioridade e quantas já são maiores.
"""

for i in range(1, 8):
    data = int(input(f"Digite o {i}º ano de nascimento: "))
    if data <= 2024-18:
        print("Maior de idade!")
    else:
        print("Ainda não atingiu a maioridade!")
```

```python
# DESAFIO 07 - V2

maior = 0
menor = 0

for i in range(1, 8):
    data = int(input(f"Digite o {i}º ano de nascimento: "))
    if data <= 2024-18:
        maior += 1
    else:
        menor += 1
print(f"\nMaiores de idade: {maior} \nMenores de idade: {menor}")
```

```python
# DESAFIO 07 - V3

maior = 0
menor = 0
anoAtual = 2024
for i in range(1, 8):
    nascimento = int(input(f"Digite o {i}º ano de nascimento: "))
    if anoAtual-nascimento >= 18:
        maior += 1
    else:
        menor += 1

print(f"Maiores: {maior}")
print(f"Menores: {menor}")
```

```python
# DESAFIO 07 - V4

anoAtual = 2024
maior    = 0
menor    = 0
for i in range(1, 8):
    idade = anoAtual-int(input(f"Digite o {i}º ano de nascimento: "))
    if idade >= 18:
        maior+=1
    else:
        menor+=1

print(f"\nMaiores de idade cadastrados: {maior}")
print(f"Menores de idade cadastrados: {menor}")
```

## Desafio 08

```python
# DESAFIO 08
"""
Faça um programa que leia um número inteiro e diga se ele é ou não um numero primo.
"""

cont=0
numero = int(input("Digite um número: "))
for i in range(1, numero):
    if numero%i == 0:
        cont+=1
    if cont > 1:
        break
if cont == 1:
    print("É primo!")
else:
    print("Não é primo!")
```

```python
# DESAFIO 08 - V2

primo = True
numero = int(input("Digite um número inteiro maior que 1: "))
for i in range(2, numero):
    if numero%i == 0:
        primo = False
        break

if primo:
    print(f"\033[32m{numero} é primo")
else:
    print(f"\033[31m{numero} não é primo")
```

## Desafio 09

```python
# DESAFIO 09
"""
Faça um programa que leia o peso de cinco pessoas. No final, mostre qual foi o maior e o menor peso lidos.
"""

maior = 0
menor = float('inf') # 'float('inf')' declara um número infinito.

for i in range(1, 6):
    peso = float(input(f"Digite o {i}º peso: "))
    if peso < menor:
        menor = peso
    else:
        maior = peso
print(f"\nMaior peso: {maior:,.2f} \nMenor peso: {menor:,.2f}")
```

```python
# DESAFIO 09 - V2

maior = 0
menor = float('inf')

for i in range(5):
    peso = float(input(f"Digite o {i+1}º peso: "))
    if peso < menor:
        menor = peso
    if peso > maior:
        maior = peso
print(f"\nMaior peso: {maior:,.2f} \nMenor peso: {menor:,.2f}")
```
