# Aula 04 - IF e ELSE

```python
# Estrutura IF
"""
OPERADORES RELACIONAIS (COMPARAÇÃO)

>  MAIOR
<  MENOR
>= MAIOR OU IGUAL
<= MENOR OU IGUAL
== IGUAL
!= DIFERENTE
"""

x = 8

if x > 5:
    print("É maior que 5")
    if x < 8:
        print("E também é menor que 8")
if x == 10:
    print("É igual a 10")

# Estrutura IF ELSE
print('-' * 30, '\n')

if x != 8:
    print("É diferente")
else:
    print("É igual")

# Estrutura IF ELIF ELSE
print('-' * 30, '\n')

if x <= 10:
    print("Frio!")
elif x <= 20:
    print('Morno!')
elif x <= 30:
    print("Quente!")
else:
    print("Muito quente!")
```

```python
# Operadores Lógicos
"""
    and = tudo tem que ser verdadeiro
    or = apenas um precisa ser verdadeiro
    not = inverte o valor
    in = tem em
"""

x = 16

# AND
if x >= 10 and x <= 20:
    print("X está entre 10 e 20")
else:
    print("X não está entre 10 e 20")
print('-'*30)

# OR
if x >= 10 or x <= 20:
    print("X é maior ou igual a 10 OU é menor ou igual a 20")
else:
    print("X não é maior ou igual a 10 OU não é menor ou igual a 20")
print('-'*30)

# IN
escola = "SENAI Ary Torres"
if 'Ary' in escola:
    print("Tem Ary")
else:
    print("Não tem Ary")
print('-'*30)

# NOT IN
if 'Ary' not in escola:
    print("Não tem Ary")
else:
    print("Tem Ary")
```

```python
# DESAFIO 09
"""
Faça um programa que leia três números e mostre qual é o
maior e qual é o menor.
"""

a = int(input("Digite 1° número: "))
b = int(input("Digite 2° número: "))
c = int(input("Digite 3° número: "))

if a > b and a > c:
    print(f"\n{a} é o maior número")
elif b > a and b > c:
    print(f"\n{b} é o maior número")
else:
    print(f"\n{c} é o maior número")

if a < b and a < c:
    print(f"{a} é o menor número")
elif b < a and b < c:
    print(f"{b} é o menor número")
else:
    print(f"{c} é o menor número")
```

```python
# DESAFIO 09 - V2

a = int(input("Digite 1° número: "))
b = int(input("Digite 2° número: "))
c = int(input("Digite 3° número: "))

if a > b and a > c:
    maior = a
elif b > a and b > c:
    maior = b
else:
    maior = c

if a < b and a < c:
    menor = a
elif b < a and b < c:
    menor = b
else:
    menor = c

print(f"\nEsse é o maior número: {maior}")
print(f"\nEsse é o menor número: {menor}")
```

```python
# DESAFIO 09 - V3

n1 = int(input("Digite o 1º número: "))
n2 = int(input("Digite o 2º número: "))
n3 = int(input("Digite o 3º número: "))
menor = maior = n1

if n1 < n2 and n1 < n3:
    if n2 < n3:
        maior = n3
    else:
        maior = n2
elif n2 < n3:
    menor = n2
    if n1 < n3:
        maior = n3
else:
    menor = n3
    if n1 < n2:
        maior = n2
print(f"\nMenor: {menor}")
print(f"Maior: {maior}")
```

```python
# DESAFIO 18
"""
Desenvolva uma lógica que leia o peso e a altura de uma
pessoa, calcule seu IMC e mostre seu status, de acordo com a
tabela abaixo:

Abaixo de 18.5: Abaixo do Peso
Entre 18.5 e 25: Peso ideal
Entre 25 e 30: Sobrepeso
Entre 30 e 40: Obesidade
Acima de 40: Obesidade Mórbida
"""

peso = float(input("Insira seu peso: "))
altura = float(input("Insira sua alutra: "))
imc = peso / altura**2

if imc < 18.5:
    print(f"\nAbaixo do peso! Seu IMC é: {imc:.2f}")
elif imc > 18.5 and imc < 25:
    print(f"\nPeso ideal! Seu IMC é: {imc:.2f}")
elif imc > 25 and imc < 30:
    print(f"\nSobrepeso! Seu IMC é: {imc:.2f}")
elif imc > 30 and imc < 40:
    print(f"\nObesidade! Seu IMC é: {imc:.2f}")
else:
    print(f"\nObesidade mórbida! Seu IMC é: {imc:.2f}")
```

```python
# DESAFIO 19
"""
Crie um programa que leia uma frase qualquer e diga se ela é um palíndromo, exemplo:
APOS A SOPA
A SACADA DA CASA
A TORRE DA DERROTA
"""

frase = input("Digite uma frase: ").lower().replace(' ', ' ')

if frase == frase[::-1]:
    print(f"A frase '{frase}' é um palíndromo!")
else:
    print(f"A frase '{frase}' NÃO é um palíndromo!")
```
