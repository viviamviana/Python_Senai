# Aula 03 - Manipulação de strings

```python
# Fatiamento de strings

texto = "Curso de Python"

setimaLetra = texto[6] # "[]" um caractere
print(setimaLetra)

decimaAteDecimaQuinta = texto [9:15] # "[:]" intervalo de caracteres. Ele desconsidera a última.
print(decimaAteDecimaQuinta)

decimaAteDecimaQuintaDeDoisEmDois = texto [9:15:2] # "[num:num]" intervalo de caracteres, "[:num]" intervalo de espaçamento dos caracteres.
print(decimaAteDecimaQuintaDeDoisEmDois)

ateQuinta = texto [:5] # "[:num]" lê-se "até tal número".
print(ateQuinta)

aPartirDaDecima = texto [9:] # "[num:]" lê-se "de tal número adiante".
print(aPartirDaDecima)

inverterTexto = texto [15:8:-1] # inverte um trecho do texto.
print(inverterTexto)

nome = "Viviam"

inverterNome = nome [::-1] # assim inverte o texto todo.
print(inverterNome)
```

```python
# Análise de strings

frase = "SENAI Ary Torres"

# Tamanho
tamanho = len(frase) # conta quantas letras tem considerando os espaços.
print(tamanho)

# Contar caracteres
contar = frase.count('r') # conta a quantidade de caracteres que está no parênteses que estiver na variável.
print(contar) # Ele diferencia maiúscula de minúscula.

# Encontrar índice (retorna -1 se não encontrar)
encontrar = frase.find("Torres") # acha a posição do primeiro caractere indicado dentro dos parênteses.
print(encontrar) # para strings, prefira utilizar o .find()

# Encontrar índice
encontrarUltima = frase.rfind("r") # acha a posição do último caractere dentro dos parênteses.
print(encontrarUltima)

# Encontrar índice (pode gerar erro)
encontrar2 = frase.index("Torres") # também acha a posição do primeiro caractere da palavra.
print(encontrar2) # Retorna um ValueError se não encontrar.

# Verificar existência
verificar = 'Ary' in frase # teste lógico se existe o que foi declarado dentro da frase.
print(verificar)
```

```python
# Análise do valor da variável

x = 'x9'

print(x.isnumeric()) # é numérico?
print(x.isalpha()) # é alfabético?
print(x.isalnum()) # é alfanumérico?
print(x.isdecimal()) # é decimal?
print(x.isspace()) # é espaço?
# retornam valores booleanos
```

```python
# Transformação de string

maiusculo = frase.upper() # transforma em maiúsculo.
print(maiusculo)

minusculo = frase.lower() # transforma em minúsculo.
print(minusculo)

capitalizar = frase.capitalize() # deixa a primeira letra da frase em maiúsculo.
print(capitalizar)

titulo = frase.title() # transforma a primeira letra de cada palavra em maiúsculo.
print(titulo)
```

```python
# Transformação de string

substituir = frase.replace('SENAI','ESCOLA') # necessário colocar a palavra que deseja substituir e a nova palavra.
print(substituir)

limpar = frase.strip() # limpa todos os espaços inúteis, como os espaços no começo e no final de cada frase.
print(limpar)

novaFrase = "   Escola Ary Torres   "
limparDireita = novaFrase.rstrip() # limpa espaços à direita.
print(limparDireita)

limparEsquerda = novaFrase.lstrip() # limpa espaços à esquerda.
print(limparEsquerda)

frase3 = "Escola Ary Torres"
print(frase3.rstrip('s')) # retira a última letra
print(frase3.lstrip('E')) # retira a primeira letra
```

```python
# Divisão e junção de strings

frase = "SENAI Ary Torres"

juntar = '-'.join(frase) # junta os caracteres com o que tiver dentro dos parênteses.
print(juntar)

dividir = frase.split() # divide as palavras em lista.
print(dividir)

print(dividir[1]) # acessa item da lista.
```

```python
# DESAFIO 01

"""
Crie um programa que leia o nome completo de uma pessoa e mostre:
 	O nome com todas as letras maiúsculas
 	O nome com todas as letras minúsculas
 	Quantas letras ao todo (sem considerar espaços)
 	Quantas letras tem o primeiro nome
"""

nome = input("Insira seu nome completo: ")

maiusculo = nome.upper()
print(f"Seu nome em maiúsculo: {maiusculo}")
minusculo = nome.lower()
print(f"Seu nome em minúsculo: {minusculo}")

nomeSemEspaco = nome.replace(' ', '')
print("Quantidade total de letras: ", len(nomeSemEspaco))

dividir = nome.split()
print("Quantas letras há no primeiro nome: ", len(dividir[0]))
```

```python
# DESAFIO 02
"""
Faça um programa que leia um numero de quatro dígitos entre 1000 a 9999 e mostre na tela
cada um dos dígitos separados.
Ex: Digite um numero: 1979

Unidade: 9
Dezena: 7
Centena: 9
Milhar: 1
"""""

n = str(input("Insira um número inteiro entre 1000 e 9999: "))

unidade = n[0]
print(f"Milhar: {unidade}")
dezena = n[1]
print(f"Centena: {dezena}")
centena = n[2]
print(f"Dezena: {centena}")
milhar = n[3]
print(f"Unidade: {milhar}")
```

## Desafio 03

```python
# DESAFIO 03
"""
Faça um programa que leia uma frase pelo teclado e mostre:

Quantas vezes aparece a letra "A"
Em que posição ela aparece a primeira vez
Em que posição ela aparece a ultima vez
"""

frase = input("Digite uma frase: ")

letraA = frase.count('a')
primeira = frase.find('a')
ultima = frase.rfind('a')
print(f"{letraA}\n{primeira}\n{ultima}")
```

```python
# DESAFIO 03 - V2

frase = input("Digite uma frase: ").lower() # tudo o que for digitado dentro do input, será transformado em minúsculo.
print(frase.count('a'))
print(frase.find('a')+1) # +1 para deixar visível para o usuário (o primeiro caracter não será mais 0 e sim 1)
print(frase.rfind('a')+1)
```

```python
# DESAFIO 03 - V3

frase = input("Digite uma frase: ").lower()
print(f"Letras 'a' encontradas: {frase.count('a')}")
print(f"1ª letra a: {frase.find('a')+1}ª posição")
print(f"Última letra a: {frase.rfind('a')+1}ª posição")
```

```python
# DESAFIO 03 - BÔNUS
"""
Refazer o DESAFIO 03 escolhendo a letra para contar e desconsiderar espaços.
"""

frase = input("Digite uma frase: ").lower()
frase = frase.replace(' ', '')
letra = input("Digite uma letra: ").lower()
print(f"\nLetras '{letra}' encontradas: {frase.count(letra)}")
print(f"1ª letra {letra}: {frase.find(letra)+1}ª posição")
print(f"Última letra {letra}: {frase.rfind(letra)+1}ª posição")
```

```python
# DESAFIO 03 - BÔNUS V2

frase = input("Digite sua frase: ").upper().replace(' ', '')
letra = input("Digite a letra: ").upper()
print(f"\nQuantidade de {letra}: {frase.count(letra)}")
print(f"1ª posição de {letra}: {frase.find(letra)+1}")
print(f"Última posição de {letra}: {frase.rfind(letra)+1}")
```
